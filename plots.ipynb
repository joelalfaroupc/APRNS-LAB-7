```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your CSV file (change the path to your file)
df = pd.read_csv("/home/oriol/IA-Universitat/APRNS/APRNS-LAB-7/ddpg.csv")

# Plot
plt.figure(figsize=(10,6))

plt.plot(df["Episode"], df["Episode Reward"], label="Reward")
plt.plot(df["Episode"], df["Mean R"], label="Mean R")

plt.xlabel("Episode")
plt.ylabel("Values")
plt.title("Reward and Mean R vs Episode")
plt.legend()
plt.axhline(0)
plt.tight_layout()
plt.show()
```
