# Archived DDPG and TD3 evaluations
Run this notebook from the repository root after installing `requirements.txt`.
`Episode Reward` is the rounded mean of three deterministic evaluation episodes; `Mean R` is the original rolling mean of up to 100 unrounded evaluation means.
There is one archived run per algorithm. Seeds were not recorded, so this is a descriptive comparison.

```python
import csv
import matplotlib.pyplot as plt

with open("ddpg.csv", newline="") as handle:
    rows = list(csv.DictReader(handle))
df = {key: [float(row[key]) for row in rows] for key in rows[0]}

plt.figure(figsize=(10, 5))
plt.plot(df["Episode"], df["Episode Reward"], alpha=0.3, label="Evaluation mean (3 episodes, rounded)")
plt.plot(df["Episode"], df["Mean R"], label="Rolling mean (up to 100 evaluations)")
plt.xlabel("Evaluation index (original zero-based numbering)")
plt.ylabel("Episode return")
plt.title("DDPG — archived evaluations")
plt.legend()
plt.axhline(0, color="gray", linewidth=0.8)
plt.tight_layout()
plt.show()
```
